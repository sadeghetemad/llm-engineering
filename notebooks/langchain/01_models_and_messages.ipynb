# 01 - Models and Messages

Covers: calling a chat model, streaming, provider-agnostic initialization, model capabilities, and message roles.

Setup: copy `.env.example` to `.env` and set `GROQ_API_KEY`.

## Calling a provider class directly

```python
from langchain_groq import ChatGroq

from llm_engineering.config import require_env

groq_llm = ChatGroq(model="openai/gpt-oss-120b", api_key=require_env("GROQ_API_KEY"))
```

```python
response = groq_llm.invoke("What is the tallest building in the world?")
response.content
```

## Streaming

`stream()` yields chunks as they are generated instead of waiting for the full answer.

```python
for chunk in groq_llm.stream("What is the tallest building in the world?"):
    print(chunk.content, end="")
```

## Provider-agnostic initialization

`init_chat_model` picks the provider from a `"<provider>:<model>"` string, so you can switch models without changing imports. The `get_llm()` helper in `llm_engineering` wraps it.

```python
from llm_engineering import get_llm

llm = get_llm()
```

```python
response = llm.invoke("What is the tallest building in the world?")
response.content
```

## Model capabilities

```python
print("Tool calling supported:", llm.profile.get("tool_calling"))
print("Max input tokens:", llm.profile.get("max_input_tokens"))
```

## Messages

Roles: `system`, `user`, `assistant`, `tool`.
Types: `SystemMessage`, `HumanMessage`, `AIMessage`, `ToolMessage`.

```python
from langchain.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage("You are a math tutor who provides answers with a bit of sarcasm."),
    HumanMessage("What is the square of 2?"),
]

response = llm.invoke(messages)  # the response is an AIMessage
response.content
```
